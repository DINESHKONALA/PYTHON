```python
import cv2
import matplotlib.pyplot as plt

```

```python
face_classifier = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
```

```python
photo = cv2.imread("photo.webp")
photo_rgb = cv2.cvtColor(photo, cv2.COLOR_BGR2RGB)
plt.imshow(photo_rgb)
plt.axis('off')
plt.show()
```

```python
gray_photo = cv2.cvtColor(photo, cv2.COLOR_BGR2GRAY)
```

```python
cv2.imwrite('gray_photo.webp', gray_photo)
photo_gray_photo = cv2.imread("gray_photo.webp")
photo_rgb_gray_photo = cv2.cvtColor(photo_gray_photo, cv2.COLOR_BGR2RGB)
plt.imshow(photo_rgb_gray_photo)
plt.axis('off')
plt.show()
```

```python
faces = face_classifier.detectMultiScale(gray_photo, scaleFactor=1.1, minNeighbors=5, minSize=(30, 30))
```

```python
for (x, y, w, h) in faces:
    cv2.rectangle(photo, (x, y), (x + w, y + h), (255, 0, 0), 2)
    cv2.imwrite('Detected_Faces.webp', photo)

photo_Detected_Faces = cv2.imread("Detected_Faces.webp")
photo_rgb_Detected_Faces = cv2.cvtColor(photo_Detected_Faces, cv2.COLOR_BGR2RGB)
plt.imshow(photo_rgb_Detected_Faces)
plt.axis('off')
plt.show()
```
