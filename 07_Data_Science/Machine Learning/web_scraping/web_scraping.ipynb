```python
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://news.ycombinator.com/"
response = requests.get(url)
print(response.text) 
# print(response.text[:500])  # Print the first 500 characters of the response
```

```python
articles = []
```

```python
soup = BeautifulSoup(response.text, 'html.parser')
for item in soup.find_all('tr', class_='athing'):
    title_tag = item.find('span', class_='titleline').find('a')
    print(title_tag.text)
    print()  # Print a newline for better readability

    title = title_tag.text
    link = title_tag.get('href' , 'N/A')

    new_article = {
        'title': title_tag.text,
        'link': title_tag['href']
    }
    articles.append(new_article)
```

```python
print(articles)
```

```python
for index , item in enumerate(soup.find_all('span' , class_='score')):
    articles[index]['upvotes'] = item.text
```

```python
print(articles) 
```

```python
url2 = 'https://books.toscrape.com/catalogue/category/books/travel_2/index.html'
response2 = requests.get(url2)
print(response2.text) 
```

```python
soup = BeautifulSoup(response2.text, 'html.parser')

books =[]

for item in soup.find_all('article', class_='product_pod'):
    title = item.find('h3').find('a').text
    price = item.find('p' , class_='price_color').text
    star_rating_class = item.find('p', class_='star-rating')
    rating = star_rating_class.get('class')[-1]


    books.append({
        'title': title,
        'price': price,
        'rating': rating,
    })
```

```python
print(books)
```
